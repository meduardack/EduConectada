```python
print("Escolas COM dado de medição válida disponivel:", educ['media_tcp_download'].notna().sum())
print("Escolas SEM dado de medição válida disponivel:", (~educ['media_tcp_download'].notna()).sum())
print("\nProporção COM dado de internet:", educ['media_tcp_download'].notna().mean())
```

```python
# Conferindo se os 15792 sem internet pelo indicador estão dentro dos 59979 sem dado de medição válida
educ['status_medicao_download'] = np.select(
    [(educ['indicador_internet'] == True) & (educ['media_tcp_download'].notna()), (educ['indicador_internet'] == True) &
     (educ['media_tcp_download'].isna()), (educ['indicador_internet'] == False) & (educ['media_tcp_download'].isna()),
      (educ['indicador_internet'] == False) & (educ['media_tcp_download'].notna())],

    ['Tem Internet declarada e medição válida', 'Tem Internet declarada, mas sem medição válida', 'Não tem Internet declarada e não tem medição válida',
      'Não tem Internet declarada, mas tem medição válida'], default='Não classificado')

resumo_medicao = (educ['status_medicao_download'].value_counts().rename_axis('Situação').reset_index(name='Quantidade de escolas'))
print(resumo_medicao)
```

```python
sem_internet = educ['indicador_internet'] == False
sem_medicao = educ['media_tcp_download'].isna()

total_sem_internet = sem_internet.sum()

sem_internet_sem_medicao = (sem_internet & sem_medicao).sum()

percentual_incluso = (sem_internet_sem_medicao / total_sem_internet * 100)
# Probabilidade Condicional

print(f'Total de escolas sem Internet declarada: 'f'{total_sem_internet:,}')

print(f'Dessas, sem medição válida de download: 'f'{sem_internet_sem_medicao:,}')

print(f'Percentual das escolas sem Internet que está dentro 'f'do grupo sem medição: {percentual_incluso:.2f}%')
```

```python
# Qual é a probabilidade de uma escola não ter Internet declarada, mas possuir medição válida de download? (dados inconsistentes)
# Probabilidade da Interseção
escolas_inconsistentes = len(educ[(educ['indicador_internet'] == False) & (educ['media_tcp_download'].notna())])
prob_inconsistencia = escolas_inconsistentes / len(educ)
print(f'Probabilidade de a escola não ter Internet declarada, mas possuir medição válida de download: {round(prob_inconsistencia*100, 2)}%')
```

```python
# Probabilidade da União
evento_consistente = (((educ['indicador_internet'] == True) &(educ['media_tcp_download'].notna())) | ((educ['indicador_internet'] == True) &
                      (educ['media_tcp_download'].isna())) | ((educ['indicador_internet'] == False) & (educ['media_tcp_download'].isna())))

prob_consistencia = evento_consistente.mean()

print(f'Probabilidade de a escola estar em uma situação consistente entre Internet declarada e medição de download:{round(prob_consistencia*100, 2)}%')
```

```python
# Dado que a escola é rural, qual a probabilidade de ter dados inconsistentes?
rurais = educ[educ['localizacao'] == 'Rural']
rurais_inconsistentes = len(rurais[(rurais['indicador_internet'] == False) & (rurais['media_tcp_download'].notna())])
prob_rural_inconsistentes = rurais_inconsistentes / len(rurais)
print(f'Dado que a escola é rural, probabilidade de ter dados inconsistentes é de: {round(prob_rural_inconsistentes*100, 2)}%')
# E consistentes?
rurais_consistentes = len(rurais[evento_consistente])
prob_rural_consistentes = rurais_consistentes / len(rurais)
print(f'Dado que a escola é rural, probabilidade de ter dados consistentes é de: {round(prob_rural_consistentes*100, 2)}%')
```

```python
# Dado que a escola é urbana, qual a probabilidade de ter dados inconsistentes?
urbanas = educ[educ['localizacao'] == 'Urbana']
urbanas_inconsistentes = len(urbanas[(urbanas['indicador_internet'] == False) & (urbanas['media_tcp_download'].notna())])
print(f'Dado que a escola é urbana, probabilidade de ter dados inconsistentes é de: {round(urbanas_inconsistentes / len(urbanas)*100, 2)}%')
# E consistentes?
urbanas_consistentes = len(urbanas[evento_consistente])
prob_urbana_consistentes = urbanas_consistentes / len(urbanas)
print(f'Dado que a escola é urbana, probabilidade de ter dados consistentes é de: {round(prob_urbana_consistentes*100, 2)}%')
```

```python
# Qual o gap entre urbana e rural na chance de ter dados consistentes?
gap = prob_urbana_consistentes - prob_rural_consistentes
print(f'O gap entre urbana e rural na probabilidade de ter dados consistentes é de: {round(gap*100, 2)} pontos percentuais')
```

```python
# Dado que a escola é rural, qual a probabilidade de ter internet declarada E medição SIMET válida?
# Probabilidade Condicional
rurais = educ[educ['localizacao'] == 'Rural']
prob_rural_com_internet_e_medicao = len(rurais[(rurais['indicador_internet'] == True) & (rurais['media_tcp_download'].notna())]) / len(rurais)
print(f'Dado que a escola é rural, probabilidade de ter internet e medição SIMET é de: {round(prob_rural_com_internet_e_medicao*100, 2)}%')
```

```python
# Dado que a escola é urbana, qual a probabilidade de ter internet declarada E medição SIMET válida?
# Probabilidade Condicional
urbanas = educ[educ['localizacao'] == 'Urbana']
prob_urbana_com_internet_e_medicao = len(urbanas[(urbanas['indicador_internet'] == True) & (urbanas['media_tcp_download'].notna())]) / len(urbanas)
print(f'Dado que a escola é urbana, probabilidade de ter internet e medição SIMET é de: {round(prob_urbana_com_internet_e_medicao*100, 2)}%')
```

As limitações observadas na base, principalmente a presença de valores nulos nas
variáveis de conectividade, nos levaram a trabalhar com dois espaços amostrais
diferentes dentro do dataset:


*   **Ω₁ = todas as ~137 mil escolas da base.**
*   **Ω₂ = subconjunto de Ω₁ com ~77 mil escolas que possuem medição SIMET válida de internet.**

Inicialmente, Ω₂ pode ser entendido apenas como um subconjunto de Ω₁. Porém,
quando mudamos a pergunta e o foco da análise de “qual é o perfil das escolas da base?” para “a velocidade de Internet é suficiente para as necessidades da escola?”, Ω₂ passa a funcionar como um novo espaço amostral que possui as informações necessárias para trabalhar comparações sobre conectividade.

Quando focamos nas variáveis de conectividade (célula 18 - 21), identificamos 1.576 escolas classificadas como sem Internet no indicador cadastral, mas que possuem uma medição válida de download pelo SIMET. Esses casos representam uma
divergência entre as informações disponíveis na base.

Vamos optar por excluir esses dados divergentes. O recorte Ω2 já é uma seleção baseada nos dados existentes de media_tcp_download. Portanto, a possibilidade de viés de seleção já existe a partir do momento em que escolhemos analisar apenas as escolas com medição válida. Excluir as 1.576 inconsistentes não elimina essa limitação e também não deve aumentá-la de forma relevante.

Então seguimos assim:
* **educ: base completa, Ω₁, 137.914 escolas.**
* **educ_ea2: recorte final e coerente para comparar atendimento, 76.359**

E direcionamos a análise para:
1. **comparar escolas urbanas e rurais em relação à conectividade,**

2. **verificar se a velocidade de download efetivamente medida parece suficiente para a necessidade estimada de cada escola.**

O foco agora é entender a diferença entre a velocidade entregue, a demanda calculada para uso pedagógico e os perfis de escola que ficam mais expostos a déficits de conexão.

Trazemos essa comparação como referência para reforçar que tanto Ω₁ quanto Ω₂

Antes de comparar grupos, precisamos entender como as velocidades se distribuem em Ω₂. Essa etapa é importante porque médias isoladas podem esconder uma realidade muito desigual: poucas escolas com velocidades muito altas podem elevar a média, mesmo quando a maior parte das escolas está concentrada em faixas mais baixas.

```python
# Entendendo melhor o cenário de download e upload.
plt.figure(figsize=(10, 6))

sns.boxplot(data=educ_ea2[['media_tcp_download', 'media_tcp_upload']], orient='h', color='skyblue')

plt.title('Distribuição das Velocidades Médias — Ω2')
plt.xlabel('Velocidade média (Mbps)')
plt.ylabel('Métrica')
plt.show()
```

```python
from scipy.stats import expon, logistic, norm

download = educ_ea2['media_tcp_download'].dropna()

param_expon = expon.fit(download)
param_logistic = logistic.fit(download)
param_norm = norm.fit(download)

x = np.linspace(download.min(), download.max(), 1000)

linha_expon = expon.pdf(x, *param_expon)
linha_logistic = logistic.pdf(x, *param_logistic)
linha_norm = norm.pdf(x, *param_norm)

fator = len(download) * (download.max() - download.min()) / 50

fig, eixos = plt.subplots(1, 2, figsize=(16, 6))


sns.histplot(
    data=educ_ea2,
    x='media_tcp_download',
    bins=50,
    kde=True,
    color='steelblue',
    ax=eixos[0])

eixos[0].plot(x, linha_expon * fator, color='orange', label='expon')
eixos[0].plot(x, linha_logistic * fator, color='green', label='logistic')
eixos[0].plot(x, linha_norm * fator, color='red', label='norm')

eixos[0].set_title('Distribuição da Velocidade Média de Download — Ω2')
eixos[0].set_xlabel('Média de download (Mbps)')
eixos[0].set_ylabel('Quantidade de escolas')
eixos[0].legend()
eixos[0].grid(axis='y', alpha=0.3)


sns.histplot(
    data=educ_ea2,
    x='media_tcp_download',
    bins=50,
    kde=True,
    color='steelblue',
    ax=eixos[1])

eixos[1].plot(x, linha_expon * fator, color='orange', label='expon')
eixos[1].plot(x, linha_logistic * fator, color='green', label='logistic')
eixos[1].plot(x, linha_norm * fator, color='red', label='norm')

eixos[1].set_xlim(0, 1000)

eixos[1].set_title('Download — Região de Maior Concentração em Ω2')
eixos[1].set_xlabel('Média de download (Mbps)')
eixos[1].set_ylabel('Quantidade de escolas')
eixos[1].legend()
eixos[1].grid(axis='y', alpha=0.3)


plt.suptitle('Distribuição da Velocidade Média de Download', fontsize=15)

plt.tight_layout()
plt.show()
```

```python
educ_ea2['diferenca_download'] = (educ_ea2['media_tcp_download'] - educ_ea2['velocidade_download_necessaria'])
educ_ea2['atende_binario'] = np.where(educ_ea2['diferenca_download'] >= 0, 1, 0)
```

A maioria das escolas não atende, o que é um grande ponto de atenção.

Ao comparar rural e urbana, é importante separar duas perguntas que parecem semelhantes, mas usam denominadores diferentes:


1.   A probabilidade condicional pergunta: dada uma escola rural ou urbana, qual é a chance de ela atender à necessidade?
2.   Já Bayes inverte a pergunta: dada uma escola que atende, qual é a chance de ela ser rural ou urbana?

```python
# status_atendimento por localizacao - Probabilidade Simples
tabela_atendimento_localizacao = pd.crosstab(
    educ_ea2['localizacao'],
    educ_ea2['status_atendimento'],
    normalize='index') * 100

print(tabela_atendimento_localizacao[['Atende', 'Não atende']].round(2))
```

```python
# Dado que a escola atende a necessidade estimada, qual a probabilidade dela ser Rural?
# Teorema de Bayes
escolas_rurais = educ_ea2[educ_ea2['localizacao'] == 'Rural']
p_atende_dado_rural = (len(escolas_rurais[escolas_rurais['status_atendimento'] == 'Atende']) / len(escolas_rurais))
p_ser_rural = (len(escolas_rurais) / len(educ_ea2))
p_atende_geral = (len(educ_ea2[educ_ea2['status_atendimento'] == 'Atende']) / len(educ_ea2))

p_rural_dado_atende = ((p_atende_dado_rural * p_ser_rural) / p_atende_geral)
print(f'Dado que a escola atende a necessidade estimada, a probabilidade dela ser rural é de: {round(p_rural_dado_atende*100,2)}%')
```

```python
# Dado que a escola atende a necessidade estimada, qual a probabilidade dela ser Urbana?
# Teorema de Bayes
escolas_urbanas = educ_ea2[educ_ea2['localizacao'] == 'Urbana']
p_atende_dado_urb = (len(escolas_urbanas[escolas_urbanas['status_atendimento'] == 'Atende']) / len(escolas_urbanas))
p_ser_urbana = (len(escolas_urbanas) / len(educ_ea2))
p_urbana_dado_atende = ((p_atende_dado_urb *p_ser_urbana) / p_atende_geral)
print(f'Dado que a escola atende a necessidade estimada, a probabilidade dela ser urbana é de: {round(p_urbana_dado_atende*100,2)}%')
```

Embora a maior parte das escolas que atendem, em números absolutos, seja urbana, isso não significa que a escola urbana tenha maior probabilidade de atender. Quando analisamos cada localização separadamente, **30,17%** das escolas rurais atendem à necessidade estimada, contra **21,64%** das urbanas.

Portanto, temos duas afirmações que são verdadeiras ao mesmo tempo:
- entre as escolas que atendem, há mais urbanas, porque há muitas urbanas em Ω₂;
- dentro do próprio grupo, a proporção de atendimento é maior nas rurais.

Essa aparente vantagem rural não deve ser interpretada como melhor qualidade de conexão rural, já negamos isso anteriormente. Então, como a velocidade necessária depende do número de estudantes que poderiam utilizar a conexão simultaneamente, o porte da escola pode estar influenciando esse resultado.

```python
educ_ea2['porte_escola'].unique()
```

```python
# Probabilidade Condicional vs Bayes:
# Dada uma escola rural ou urbana de determinado porte, qual a chance de ela atender à necessidade?
tabela_probabilidade_simples_porte = pd.crosstab([educ_ea2['porte_escola'], educ_ea2['localizacao']], educ_ea2['status_atendimento'], normalize='index') * 100
print(tabela_probabilidade_simples_porte[['Atende', 'Não atende']].round(2))
# Dado que a escola pertence a um determinado porte, qual a chance de ela ser rural ou urbana no quesito atendimento?
portes_comparaveis = educ_ea2['porte_escola'].dropna().unique()
tabela_bayes_porte = pd.crosstab([educ_ea2['porte_escola'], educ_ea2['status_atendimento']], educ_ea2['localizacao'], normalize='index') * 100
print(tabela_bayes_porte.reindex(portes_comparaveis, level='porte_escola').round(2))
```

```python
# Gráfico Bayes
bayes_atende_porte = (tabela_bayes_porte.xs('Atende', level='status_atendimento').reindex(portes_comparaveis))
ax = bayes_atende_porte.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    color=['#E69F00', '#56B4E9'],
    edgecolor='black')

ax.set_title('Localização das escolas que atendem, por porte — Ω2', fontsize=14)

ax.set_xlabel('Porte da escola')
ax.set_ylabel('Percentual entre as escolas que atendem')

ax.legend(title='Localização')
plt.xticks(rotation=25)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# As escolas rurais estão mais concentradas em faixas pequenas, certo?
ordem_porte = [
    'Menos de 50',
    '51 a 200',
    '201 a 500',
    '501 a 1000',
    '1001 a 2000',
    'Mais de 2000'
]

funcao_probabilidade_porte = pd.crosstab(
    educ_ea2['porte_escola'],
    educ_ea2['localizacao'],
    normalize='columns'
).reindex(
    index=ordem_porte,
    columns=['Rural', 'Urbana'],
    fill_value=0
)

print('Função de probabilidade do porte por localização:')
print((funcao_probabilidade_porte * 100).round(2))
```

```python
funcao_reparticao_porte = funcao_probabilidade_porte.cumsum()

print('Função repartição acumulada do porte por localização:')
print((funcao_reparticao_porte * 100).round(2))
```

```python
plt.figure(figsize=(10, 6))

plt.step(
    funcao_reparticao_porte.index,
    funcao_reparticao_porte['Rural'],
    where='post',
    marker='o',
    linewidth=2.5,
    color='#E69F00',
    label='Rural'
)

plt.step(
    funcao_reparticao_porte.index,
    funcao_reparticao_porte['Urbana'],
    where='post',
    marker='o',
    linewidth=2.5,
    color='#56B4E9',
    label='Urbana'
)

plt.title(
    'Função repartição acumulada do porte escolar por localização — Ω2',
    fontsize=14
)

plt.xlabel('Porte da escola')
plt.ylabel('Probabilidade acumulada')

plt.xticks(rotation=25)
plt.ylim(0, 1.05)
plt.legend(title='Localização')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

A função de probabilidade mostra, para cada localização, a participação de cada faixa de porte no próprio grupo. A função repartição acumula essas probabilidades e responde: qual a proporção de escolas rurais ou urbanas com porte menor ou igual a determinada faixa?

O gráfico mostra que em Ω2, escolas rurais concentram-se nas faixas de menor porte, enquanto escolas urbanas aparecem com maior frequência nas faixas médias e grandes. Como a velocidade necessária é estimada conforme a capacidade de atender 50% dos alunos simultaneamente, essa diferença de porte ajuda a contextualizar a maior necessidade de velocidade nas escolas urbanas e a aparente contradição observada no indicador de atendimento.

## Normalização

Preparação para um possível modelo de Machine Learning

Depois de observar a relação entre porte da escola, velocidade necessária e
velocidade entregue, vamos imaginar que essas variáveis serão utilizadas em
um modelo de classificação.

Como elas possuem escalas diferentes, uma variável pode acabar influenciando
o modelo apenas por apresentar números maiores. Por isso, vamos comparar duas
formas de normalização: StandardScaler e MinMaxScaler.

Foram escolhidas `quantidade_matricula`, `velocidade_download_necessaria` e
`media_tcp_download` porque elas representam, respectivamente, o porte da
escola, a demanda estimada de conectividade e a velocidade efetivamente
entregue.

```python
variaveis_modelo = ['quantidade_matricula', 'velocidade_download_necessaria', 'media_tcp_download']

dados_modelo = educ_ea2[variaveis_modelo].copy()
dados_modelo.info()
```

```python
dados_modelo.isnull().sum()
```

```python
dados_modelo.describe()
```

```python
educ_ea2['quantidade_matricula'].plot(kind='hist', bins=20, title='Distribuição Matriculas')
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

#StandardScaler
scaler_standard = StandardScaler()
dados_standard = scaler_standard.fit_transform(dados_modelo)
dados_scale = pd.DataFrame(
dados_standard,
columns=variaveis_modelo,
index=dados_modelo.index)

print('Resumo após StandardScaler:')
display(dados_scale.describe())
```

```python
dados_scale['quantidade_matricula'].plot(kind='hist', bins=20, title='Distribuição Matriculas')
plt.show()
```

```python
#MinMaxScaler
scaler_minmax = MinMaxScaler()
dados_minmax_array = scaler_minmax.fit_transform(dados_modelo)
dados_minmax = pd.DataFrame(
dados_minmax_array,
columns=variaveis_modelo,
index=dados_modelo.index)
print('Resumo após MinMaxScaler:')
display(dados_minmax.describe())
```

```python
dados_minmax['quantidade_matricula'].plot(kind='hist', bins=20, title='Distribuição Matriculas')
plt.show()
```

```python
fig, eixos = plt.subplots(
    len(variaveis_modelo),
    3,
    figsize=(18, 12)
)

for linha, variavel in enumerate(variaveis_modelo):

    # Dados originais
    eixos[linha, 0].hist(
        dados_modelo[variavel],
        bins=30,
        color='gray',
        edgecolor='black'
    )

    eixos[linha, 0].set_title(
        f'{variavel} — Original'
    )

    eixos[linha, 0].set_xlabel('Valor original')
    eixos[linha, 0].set_ylabel('Quantidade')
    eixos[linha, 0].grid(axis='y', alpha=0.3)

    # StandardScaler
    eixos[linha, 1].hist(
        dados_scale[variavel],
        bins=30,
        color='royalblue',
        edgecolor='black'
    )

    eixos[linha, 1].set_title(
        f'{variavel} — StandardScaler'
    )

    eixos[linha, 1].set_xlabel('Valor padronizado')
    eixos[linha, 1].set_ylabel('Quantidade')
    eixos[linha, 1].grid(axis='y', alpha=0.3)

    # MinMaxScaler
    eixos[linha, 2].hist(
        dados_minmax[variavel],
        bins=30,
        color='darkorange',
        edgecolor='black'
    )

    eixos[linha, 2].set_title(
        f'{variavel} — MinMaxScaler'
    )

    eixos[linha, 2].set_xlabel('Valor entre 0 e 1')
    eixos[linha, 2].set_ylabel('Quantidade')
    eixos[linha, 2].grid(axis='y', alpha=0.3)

plt.suptitle(
    'Comparação entre dados originais, StandardScaler e MinMaxScaler',
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
fig, eixos = plt.subplots(1, 2, figsize=(14, 5))

dados_scale[variaveis_modelo].boxplot(ax=eixos[0])

eixos[0].set_title('Dados após StandardScaler')
eixos[0].set_ylabel('Valores padronizados')
eixos[0].tick_params(axis='x', rotation=25)
eixos[0].grid(axis='y', alpha=0.3)

dados_minmax[variaveis_modelo].boxplot(ax=eixos[1])

eixos[1].set_title('Dados após MinMaxScaler')
eixos[1].set_ylabel('Valores entre 0 e 1')
eixos[1].tick_params(axis='x', rotation=25)
eixos[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

Os gráficos mostram que a normalização não altera o formato geral das
distribuições. Ela apenas modifica a escala dos valores.

O StandardScaler centraliza os dados em torno de 0 e transforma o desvio
padrão em aproximadamente 1. Assim, os valores passam a representar a
distância de cada observação em relação à média.

O MinMaxScaler transforma os valores para o intervalo entre 0 e 1. Porém,
como as variáveis apresentam valores extremos, grande parte das observações
fica concentrada em uma faixa pequena.

Para este cenário de classificação, escolhemos o StandardScaler como ponto de
partida, pois ele preserva melhor a noção de distância em relação à média.
Essa escolha é adequada para a etapa exploratória, mas em um projeto real
também seria necessário avaliar o efeito dos outliers e o desempenho efetivo
do modelo.

## Distribuição Amostral

A variável escolhida será `media_tcp_download`.

Serão simuladas amostras com diferentes tamanhos. Para cada amostra,
calcularemos sua média e observaremos a distribuição dessas médias.

```python
from numpy.random import default_rng

populacao = educ_ea2['media_tcp_download'].dropna()

rng = default_rng(42)

media_populacao = populacao.mean()
desvio_populacao = populacao.std()
tamanho_populacao = len(populacao)

print(f'Tamanho da população: {tamanho_populacao}')
print(f'Média da população: {media_populacao:.2f} Mbps')
print(f'Desvio padrão da população: {desvio_populacao:.2f} Mbps')
```

```python
from numpy.random import seed

seed(42)

velocidades = educ_ea2['media_tcp_download'].dropna().values

def medias_amostrais(n, repeticoes=1000):return [np.mean(np.random.choice(velocidades, n, replace=True))for _ in range(repeticoes)]

tamanhos = [1, 5, 10, 30, 100]
medias_por_tamanho = {}

for n in tamanhos:
    medias_por_tamanho[n] = medias_amostrais(n)

for n in tamanhos:

    medias = medias_por_tamanho[n]

    plt.figure(figsize=(8, 5))

    plt.hist(
        medias,
        bins=30,
        color='mediumpurple',
        edgecolor='black',
        alpha=0.8
    )

    plt.axvline(
        media_populacao,
        color='red',
        linestyle='--',
        linewidth=2,
        label='Média da população'
    )

    plt.title(
        f'Distribuição das médias amostrais — n = {n}'
    )

    plt.xlabel('Velocidade média amostral (Mbps)')
    plt.ylabel('Quantidade de simulações')
    plt.legend()
    plt.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    plt.show()
```

```python
resumo_amostral = pd.DataFrame({
    'Tamanho da amostra': tamanhos,

    'Média das médias amostrais': [
        np.mean(medias_por_tamanho[n])
        for n in tamanhos
    ],

    'Desvio padrão das médias': [
        np.std(medias_por_tamanho[n], ddof=1)
        for n in tamanhos
    ],

    'Erro padrão teórico': [
        desvio_populacao / np.sqrt(n)
        for n in tamanhos
    ],

    'Mínimo': [
        np.min(medias_por_tamanho[n])
        for n in tamanhos
    ],

    'Máximo': [
        np.max(medias_por_tamanho[n])
        for n in tamanhos
    ]
})

display(resumo_amostral.round(2))
```

```python
dados_boxplot = [
    medias_por_tamanho[n]
    for n in tamanhos
]

plt.figure(figsize=(12, 6))

plt.boxplot(
    dados_boxplot,
    labels=[f'n = {n}' for n in tamanhos]
)

plt.axhline(
    media_populacao,
    color='red',
    linestyle='--',
    linewidth=2,
    label='Média da população'
)

plt.title(
    'Distribuição das médias amostrais por tamanho de amostra'
)

plt.xlabel('Tamanho da amostra')
plt.ylabel('Velocidade média amostral (Mbps)')
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

Os histogramas mostram que, para uma amostra de tamanho 1, as médias amostrais
são muito dispersas, pois cada média corresponde praticamente à velocidade de
uma única escola.

À medida que o tamanho da amostra aumenta, as médias ficam mais concentradas
em torno da média da população. O boxplot confirma essa redução da dispersão:
as caixas ficam progressivamente menores conforme o tamanho da amostra aumenta.

A tabela também mostra que a média das médias amostrais permanece próxima da
média populacional. Já o desvio padrão das médias diminui com o aumento de n,
o que indica que amostras maiores produzem estimativas mais estáveis.

Portanto, os resultados mostram que o tamanho da amostra influencia a
confiabilidade da estimativa. Amostras pequenas podem representar realidades
muito diferentes dependendo das escolas selecionadas; amostras maiores tendem
a produzir uma média mais estável da velocidade de download.
